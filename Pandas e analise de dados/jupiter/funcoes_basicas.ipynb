```python
#Uma maneira de filtra os dados
df[((df.education == 'secondary') & (df.age >= 50)) & (df.contact == "cellular")].loc[0:200]
```

```python
#Outra forma de filtrar o data frame usando isin, porem quando usamos ele diretamente no dataFrame ele retorna uma lista de bool
df_filtrado = df.isin({"marital":["married"],"education":["secondary"], "contact":["cellular"], "job":["management", "blue-collar"]})

df[(df_filtrado.education) & (df_filtrado.contact) & (df_filtrado.job) & (df_filtrado.marital)]
```

```python
#Podemos agrupar os valores por calunas, neste comando pegamos os valores da coluna marital e seu respectivos valores da coluna age e calculamos a media dos valores
#como a coluna marital tem 3 valores possíveis tivemos apenas 3 linhas e suas respectivas medias, mas não oq aconteciria se uma coluna tivesse muitos valores possíveis
df[["marital", "age"]]. groupby("marital").mean()
```
