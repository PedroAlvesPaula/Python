```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv(".//..//data//houses_to_rent.csv", sep=",")
df_nao_tratado = pd.read_csv(".//..//data//houses_to_rent.csv", sep=",")
df.head()
```

```python
df["rent amount_tratado"] = df["rent amount"].apply(lambda x: x.replace("R$", "")).apply(lambda x: x.replace(",",""))
df["ptoperty tax_tratado"] = df["property tax"].apply(lambda x: x.replace("R$", "")).apply(lambda x: x.replace(",",""))
df["fire insurance_tratado"] = df["fire insurance"].apply(lambda x: x.replace("R$", "")).apply(lambda x: x.replace(",",""))
df["total_tratado"] = df.total.apply(lambda x: x.replace("R$", "")).apply(lambda x: x.replace(",",""))


del df["rent amount"]
del df["property tax"]
del df["fire insurance"]
del df["total"]
```

```python
df["hoa_tratado"] = df.hoa.apply(lambda x: x.replace("R$", "")).apply(lambda x: x.replace(",",""))
del df["hoa"]

df.head()
```

```python
df_nao_tratado.head()
```

```python
#trocando nomes das colunas com rename

colunas_originais = df.columns.to_list()

colunas_traduzidas = ["indices", "cidade", "area", "quartos", "banheiros", "vagas_garagem", "andar", "permitido_animais", "mobilhado", "aluguel", "taxas", "seguro_contra_incendio", "total", "hoa"]

dict_traduzido = dict(zip(colunas_originais, colunas_traduzidas))

new_df = df.rename(columns=dict_traduzido)
```

```python
new_df["mobilhado"] = new_df["mobilhado"].apply(lambda x: x.replace("furnished", "mobilhado"))

new_df.head()
```

```python
new_df["mobilhado"] = new_df["mobilhado"].apply(lambda x: x.replace("not mobilhado", "nao_mobilhado"))
new_df
```

```python
new_df["permitido_animais"] = new_df["permitido_animais"].apply(lambda x: x.replace("acept", "sim"))
new_df["permitido_animais"] = new_df["permitido_animais"].apply(lambda x: x.replace("not sim", "nao"))

new_df
```

```python
#Podemos utilizar o assigned para inserir colunas no data frame, não altera o dataframe original, retorna uma copia.
new_df.assign(total_comodos = lambda x: x.quartos + x.banheiros).loc[:10, ["quartos", "banheiros", "total_comodos"]]
```
